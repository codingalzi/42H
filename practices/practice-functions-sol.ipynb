## 문제 3

다음 세 개의 함수는 모두 입력된 문자열이 소문자를 포함하고 있는지 여부를 조사하도록 구현되었지만 
하나만 제대로 작동한다.
어느 함수가 제대로 작동하는지 확인한 다음 그 이유를 설명하라.
또한 다른 두 함수가 적절하게 구현되지 않은 이유를 설명하라.

````python
def any_lowercase1(text):
    for c in text:
        flag = c.islower()
    return flag

def any_lowercase2(text):
    for c in text:
        if not c.islower():
            return False
    return True

def any_lowercase3(text):
    flag = False
    for c in text:
        if c.islower():
            flag = True
    return flag
````

**`chr()` 함수**

`chr()` 함수는 `ord` 함수의 역함수에 해당한다.

영어 소문자 알파벳은 총 26개이에 각각의 알파벳 `char`를 13 자리만큼 회전시키면
`ord(char)` 값에 따라 다르게 계산된다.

- `97 <= ord(char) <= 109`인 경우, 즉 영어 소문자 `'a'`부터 `'m'`까지의 알파벳인 경우: `ord(char)`에 13을 더해준 값에 해당하는 알파벳
- `109 <= ord(char) <= 122`인 경우, 즉 영어 소문자 `'n'`부터 `'z'`까지의 알파벳인 경우: `ord(char)`에서 13을 빼준 값에 해당하는 알파벳

따라서 모두 소문자로 변환한 다음에 각각의 알파벳을 13 자리만큼 회전시킨 결과로 구성된
문자열 아래 `encrypt()` 함수로 계산한다.

````python
def encrypt(text):
    encrypted = ""
    
    for char in text.lower():
        if 97 <= ord(char) <= 109:
            encrypted += chr(ord(char) + 13)
        elif 109 < ord(char) <= 122:
            encrypted += chr(ord(char) - 13)
        else:
            encrypted += char
            
    return encrypted       
````

````python
encrypt(encrypt('I love python'))
````

````python
encrypt('Believe me!')
````

**질문 2**

소문자와 대문자를 그대로 유지하도록 `encrypt()` 함수를 수정하라.

**답**

대소문자에 따라 `if ... elif ... else... ` 조건식을 사용하면 된다.
대문자의 구분은 숫자만 다를 뿐 소문자의 구분과 동일하다.

````python
def encrypt(text):
    encrypted = ""
    
    for char in text:
        if char.isupper():
            if 65 <= ord(char) <= 77:
                encrypted += chr(ord(char) + 13)
            else:
                encrypted += chr(ord(char) - 13)
        elif char.islower():
            if 97 <= ord(char) <= 109:
                encrypted += chr(ord(char) + 13)
            else:
                encrypted += chr(ord(char) - 13)
        else:
            encrypted += char
            
    return encrypted       
````

````python
encrypt('I love python')
````

````python
encrypt(encrypt('I love python'))
````

````python
encrypt('Believe me!')
````

## 문제 3

아래 코드의 실행 결과를 설명하라.
반복문 실행횟수는 얼마인가?

````python
def counter_func(num, limit) :
    i = 1
    while True :
        print(i)
        if i == limit :
            return i
        else:
            print("=" * i)
            print("*" * i)
            i += 1
        return num

counter_func(100, 5)
````

**답**

`counter_func(100, 5)`이 호출되면 함수의 본문이 차례대로 실행된다.
`i = 1` 에서 시작하고 `while True` 반복문이 실행된다. 
논리식 `True`가 참이라 `while` 반복문이 무한 반복되어야 하는 것처럼 보인다.
`print(i)`에 의해 1이 출력되고
`if ... else ...` 조건문에서 `i == limit`가 거짓이기에 `else`의 본문이 실행되어
`"="`와 `"*"` 가 출력된다.

그런데 바로 이어서 `return num` 이 실행되어 100을 반환하고 반복문을 비롯해서 
함수의 실행이 완전히 종료된다.

## 문제 5

정수 `num`을 인자로 해서 호출되었을 때, 1의 자릿수가 3의 배수이면 입력값의 세 배를, 
아니면 두 배를 반환하는 함수 `three_or_not()` 함수를 구현하라.
매개 변수는 `num` 하나만 사용한다.

힌트: `if ... else ...` 조건문과 나머지 연산자 `%` 활용

**답**

정수의 1의 자릿수는 10으로 나눴을 때의 나머지이다. 

````python
123 % 10
````

````python
27 % 10
````

3의 배수 여부는 3으로 나눈 나머지가 0이냐에 의해 결정된다.
따라서 `num`을 10으로 나눴을 때의 나머지에 대해 3의 배수 여부를 확인하고
그에 따라 `num`의 3배수 또는 2배수를 반환값으로 지정한다.

````python
def three_or_not(num):
    unit = num % 10
    if unit % 3 == 0:
        return num * 3
    else:
        return num * 2
````

````python
three_or_not(13)
````

````python
three_or_not(24)
````

## 문제 6

정수 인자와 함께 호출되어 인자가 짝수일 때는 1, 홀수일 때는 0을 반환하는 함수 `even_odd()`를 구현하라.
매개 변수는 `num` 하나만 사용한다.

힌트: `if ... else ...` 조건문과 나머지 연산자 `%` 활용

**답**

`num`의 인자로 지정되는 정수의 짝수, 홀수 여부는 2로 나눈 나머지가 0인지, 1인지에 의해 결정된다.
따라서 `num % 2 == 0` 논리식의 진릿값에 따른 `if` 조건문을 아래처럼 사용한다.

````python
def even_odd(num):
    if num % 2 == 0:
        answer = 1
    else:
        answer = 0  
      
    return answer
````

````python
even_odd(10)
````

````python
even_odd(11)
````

참고로 `True`는 정수 1로, `False`는 정수 0으로 간주되기도 한다.
형변환 함수 `int()`를 이용하여 확인할 수 있다.

````python
int(True)
````

````python
int(False)
````

이 성질을 이용하면 `even_odd()` 함수를 아래처럼 보다 간단하게 정의할 수 있다.

````python
def even_odd(num):
    return int(num % 2 == 0)
````

````python
even_odd(10)
````

````python
even_odd(11)
````

`even_odd()` 함수를 이용하여 짝수, 홀수 여부를 판단한 결과를 다른 명령문에 활용할 있다.
예를 들어, 아래 `divide_by_2()` 함수는 짝수면 2로 나누고, 홀수면 1을 더한 값을 2로 나눈다.

함수 본문의 `if ... ` 조건문의 논리식으로 굳이 정수를 부울값으로 형변화는 함수인 `bool()`을 사용할 필요는 없다.
이는 파이썬 실행기가 0은 `False`로 0이 아니면 `True`로 처리하기 때문이다.

````python
def divide_by_2(num):
    if even_odd(num):
        return num // 2
    else:
        return (num + 1) // 2
````

````python
divide_by_2(11)
````

````python
divide_by_2(12)
````

[(필수 예제) 함수](https://colab.research.google.com/github/codingalzi/pybook/blob/master/examples/examples-functions.ipynb#scrollTo=8ca4c478)에서 살펴 본 아래 함수는 
양의 정수 `n`과 함께 호출되면 `n x n` 모양의 격자무늬를 출력한다.
단, `while` 반복문 대신에 `for` 반복문과 `range()` 함수가 사용되었다.

````python
pm = "+ - - "
ps = "|     "
plus_sign = "+"
pipe_sign = "|"

def print_pm(n):
    print(pm * n + plus_sign)

def print_ps(n):
    ps_line = ps * n + pipe_sign
    print(ps_line, ps_line, sep="\n")

def grid(n):
    for count in range(n):
        print_pm(n)
        print_ps(n)
        count += 1
        
    print_pm(n)
````

````python
grid(3)
````

- `m`: 가로 격자 칸 개수
-  반환값: `n*m`

참고한다.

- 가로 격자 칸 개수: `print_pm(n)` 과 `print_ps(n)`에서의 `n`에 의해 결정됨
- 세로 격자 칸 개수: `for count in range(n)`에서의 `n`에 의해 결정됨.

따라서 가로와 세로의 격자 칸 개수를 달리 하려면 
앞서 언급된 두 부분에서의 `n`의 사용을 달리 해야 한다.
아래 코드는 `n`은 세로 격자 칸 개수를, `m`은 가로 격자 칸 개수를 지정한다.

````python
pm = "+ - - "
ps = "|     "
plus_sign = "+"
pipe_sign = "|"

def print_pm(m):              # m: 가로 격자 칸 개수
    print(pm * m + plus_sign)

def print_ps(m):
    ps_line = ps * m + pipe_sign
    print(ps_line, ps_line, sep="\n")

def grid_gen(n, m):
    for count in range(n):         # n: 세로 격자 칸 개수
        print_pm(m)
        print_ps(m)
        count += 1
        
    print_pm(m)
    
    return n * m            # 반환값: 격자 칸 수
````

````python
grid_gen(2, 3)
````

````python
grid_gen(3, 5)
````
