**예제: 리스트에 포함된 모든 항목의 합**

1부터 5까지의 합을 계산하는 코드를 `for` 반복문과 리스트를 활용하여 구현한다.
또한 다음과 같이 출력되도록 한다.

> 1부터 5까지 정수의 합: 15

모범답안:

먼저 1부터 5을 항목으로 갖는 리스트를 `one2five` 변수에 할당한다.

````python
one2five = [1, 2, 3, 4, 5]
````

리스트와 함께 사용되는 `for` 반복문은 문자열의 0번 인덱스의 항목부터 차례대로 활용한다는 사실을 이용해서
아래 형식으로 덧셈을 진행하는 코드를 작성한다.

아래 `for` 반복문은 `sum_list` 변수를 0으로 설정한 다음에
`item` 변수가 가리키는 값이 달라질 때마다 `sum_list` 변수가 가리키는 값을 업데이트한다.

```
sum_list = 0
sum_list = sum_list + item =  0 + 1 =  1
sum_list = sum_list + item =  1 + 2 =  3
sum_list = sum_list + item =  3 + 3 =  6
sum_list = sum_list + item =  6 + 4 = 10
sum_list = sum_list + item = 10 + 5 = 15
```

````python
sum_list = 0

for item in one2five:
    sum_list = sum_list + item
    
print("1부터 5까지 정수의 합:", sum_list)
````

간편한 변수 업데이트 형식을 사용하는 코드는 다음과 같다.

````python
sum_list = 0
for item in one2five:
    sum_list += item
    
print("1부터 5까지 정수의 합:", sum_list)
````
